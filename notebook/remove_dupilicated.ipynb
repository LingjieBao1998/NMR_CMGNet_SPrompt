```python
import pandas as pd
import numpy as np
import os
import rdkit
from rdkit import Chem
```

```python
df0 = pd.read_csv("../data/positive.csv")
df0
```

```python
def strard_mol_fn(smiles):
    try:
        new_smiles = Chem.MolToSmiles(Chem.MolFromSmiles(Chem.MolToSmiles(Chem.MolFromSmiles(smiles))))
        return new_smiles
    except Exception as e:
        print(e)
        return ""
```

```python
df0["Cano_SMILES"] = df0["SMILES"].apply(strard_mol_fn)
```

```python
df1 = df0[df0["Cano_SMILES"]!=""]
```

```python
len(df0), len(df1)
```

```python
df2 = df1.drop_duplicates(subset="Cano_SMILES")
```

```python
df2 = df2.reset_index(drop=True)
df2
```

```python
df2.to_csv("../data/positive_remove_duplicated.csv")
```
